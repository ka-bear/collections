### What is a Cost Function?
The point of an AI model is to predict trends. However, in the process of training, we will need to find a way to measure **how far off our prediction is** from the true values. This is also known as the **error**.  

The function to determine the error of our predictions is called a **cost function** or a **loss function** (they're the same function, just in different contexts!)  

All cost functions will provide a magnitude of how 'wrong' a prediction is. As such, we will be able to tell how close we are to the theoretical 'perfect' model - given that the data is normalized appropriately, for example to values between 0 and 1!  

### The Different Cost Functions
Recall that AI models are split into two different types: **Regression** and **Classification**. As both are fundementally different tasks, there are different Cost Functions for the two types.  

For Regression, the Cost Functions we will be covering are:
- Mean Squared Error (MSE)
- Mean Absolute Error (MAE)

For Classification, the Cost Functions we will only be covering Binary and Multiclass Cross Entropy.  

You must never use Cost Functions meant for Regression in Classification, and vice versa. Although it may(?) work, the appropriate Cost Function for your use case will work much better.

### Mean Error, Mean Squared Error, and Mean Absolute Error
These Cost Functions all stem from the same core idea - the error of a single prediction is simple `true value - predicted value`, or `y - yhat`. Tuning the model to reduce this error would result in a more accurate model. 

The Mean Error Cost Function is what you get when you sum this error across the training data, then divide it by the number of samples (which is identical to taking the mean). However, the error of each prediction can be either positive or negative. Summing positive and negative errors together will result in the errors cancelling out, which is not ideal as the Cost Function should be capturing how far off the model is in general!  
As an illustration, imagine our errors are [-5, 5, -3, 4, 1, -1]. Summing these together, we would get a Mean Error of 1/6, which is far off considering that the errors have magnitudes much greater than that!  

As such, how can we address this issue?  

#### Mean Squared Error (MSE) / L2 Loss
As the name suggests, Mean Squared Error is the mean of the square of the error. Hence, instead of summing the error then dividing it by the number of samples, we sum the square of the error, to ensure that the value is positive, then get the mean. This way, the errors will not cancel out, and we will have a fair representation of the error of the model in general!  

The equation for MSE is `sum((y - yhat)^2)/n` (imagine this in written form)  

This Cost Function is also known as the L2 Loss function.  

**Pros**:
- Square of error penalizes even small deviations from the true value
**Cons**:
- Square of error magnifies the effect of outliers! Hence MSE is less robust to outliers

How, then, can we address this issue?

#### Mean Absolute Error (MAE) / L1 Loss
Instead of squaring the error to remove the sign, we can just get the absolute magnitude of the error instead. 

The equation for MAE is `sum(|y - yhat|)/n`.

**Pro**: More robust against outliers!
